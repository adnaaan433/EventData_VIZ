```python
df = pd.read_csv(r"D:\FData\All_Top_5_League\2024_25\event_data\Top5_League_2024_25_event_data.csv")
```

```python
head_shots = df[(df['type'].isin(['Goal', 'MissedShots', 'SavedShot', 'ShotOnPost'])) & (df['qualifiers'].str.contains('Head')) & (~df['qualifiers'].str.contains('FromCorner|SetPiece')) &
                (df['position'].isin([
                    # 'AMC', 
                    'AMR', 'AML', 'FWR', 'FWL']))]
```

```python
# Group by playerName and aggregate
pdf = head_shots.groupby('playerName').agg({
    'type': 'count',  # Count occurrences
    'xG': 'sum'       # Sum xG values
}).rename(columns={'type': 'total_shots', 'xG': 'total_xG'}).reset_index()

pdf = pdf.sort_values(by='total_xG', ascending=False)
pdf = pdf[pdf['total_shots']>1]
pdf
```
