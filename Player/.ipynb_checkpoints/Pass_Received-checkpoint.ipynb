```python
def playing_time_calculator(team_name, pname):
    dfteam = df[(df['name']==pname) & (df['teamName']==team_name)]
    mids = dfteam.match_id.unique()
    total_playing_time = 0
    for mid in mids:
        dfmi = df[df['match_id']==mid]
        # Filter player data
        df_playing_time = dfmi[dfmi["name"] == pname]
        df_playing_time["isFirstEleven"] = df_playing_time["isFirstEleven"].fillna(0)
    
        # Identify substitution events
        df_sub_off = df_playing_time[df_playing_time["type"] == "SubstitutionOff"]
        df_sub_on = df_playing_time[df_playing_time["type"] == "SubstitutionOn"]
    
        # Get maximum match minute and extra time
        max_min = dfmi["minute"].max()
        extra_time = max(0, max_min - 90)  # Ensure extra_time is non-negative
    
        # Initialize minutes played
        mins_played = 0
    
        # Case 1: Started the game and was not substituted off
        if df_playing_time["isFirstEleven"].unique()[0] == 1 and len(df_sub_off) == 0:
            mins_played = 90
    
        # Case 2: Started the game and was substituted off
        elif df_playing_time["isFirstEleven"].unique()[0] == 1 and len(df_sub_off) == 1:
            sub_off_min = df_sub_off["minute"].unique()[0]
            mins_played = min(90, sub_off_min)
            if sub_off_min > 90:
                mins_played = int((sub_off_min * 90) / max_min)  # Proportional adjustment for extra time
    
        # Case 3: Substituted on before or at the 80th minute
        elif df_playing_time["isFirstEleven"].unique()[0] == 0 and len(df_sub_on) > 0:
            sub_on_min = df_sub_on["minute"].unique()[0]
            if sub_on_min <= 80:
                mins_played = max_min - sub_on_min - extra_time
            else:
                mins_played = max_min - sub_on_min
    
        # Adjust for red cards
        df_red = df_playing_time[(df_playing_time["type"] == "Card") & (df_playing_time["qualifiers"].str.contains("SecondYellow|Red", na=False))]
        if len(df_red) == 1:
            red_min = df_red["minute"].max()
            mins_played = mins_played - (90 - red_min)

        mi_playing_time = int(mins_played)
        total_playing_time = total_playing_time + mi_playing_time

    return total_playing_time
```

```python
def plot_pass_recieve(ax, team_name, pname):
    dfp = df[(df['type']=='Pass') & (df['outcomeType']=='Successful') & (df['name'].shift(-1)==pname) & (df['teamName']==team_name)]
    dfop = dfp[(~dfp['qualifiers'].str.contains('FreeKick|Corner'))]
    dfkp = dfp[dfp['qualifiers'].str.contains('KeyPass')]
    dfas = dfp[dfp['qualifiers'].str.contains('IntentionalGoalAssist')]
    dfnt = dfp[dfp['endX']>=70]
    dfpen = dfp[(dfp['endX']>=87.5) & (dfp['endY']>=13.6) & (dfp['endY']<=54.6)]
    dfpro = dfp[(dfp['x']>=35) & (dfp['prog_pass']>=9.11)]
    dfcros = dfp[(dfp['qualifiers'].str.contains('Cross'))]
    dfthr = dfp[dfp['qualifiers'].str.contains('Throughball')]
    dflb = dfp[(dfp['qualifiers'].str.contains('Longball')) & (dfp['x']<=52)]

    # Most Passes from
    name_counts = dfp['shortName'].value_counts()
    name_counts_df = name_counts.reset_index()
    name_counts_df.columns = ['name', 'count']
    name_counts_df = name_counts_df.sort_values(by='count', ascending=False)  
    name_counts_df = name_counts_df.reset_index()

    pitch = VerticalPitch(pitch_type='uefa', pitch_color='#f5f5f5', corner_arcs=True, linewidth=2)
    pitch.draw(ax=ax)
    
    kp = pitch.lines(dfkp.x, dfkp.y, dfkp.endX, dfkp.endY, color='#e322cc', lw=2.35, comet=True, transparent=True, ax=ax)
    ass = pitch.lines(dfas.x, dfas.y, dfas.endX, dfas.endY, color='green', lw=2.5, comet=True, transparent=True, ax=ax)

    ax.scatter(dfop.endY, dfop.endX, s=35, color='#f5f5f5', alpha=0.75, edgecolor='#2680ff', zorder=1)
    ax.scatter(dfkp.endY, dfkp.endX, s=35, color='#f5f5f5', edgecolor='#e322cc', zorder=3)
    ax.scatter(dfas.endY, dfas.endX, s=35, color='#f5f5f5', edgecolor='green', zorder=4)

    medX = dfp['endX'].median()
    medY = dfp['endY'].median()
    
    # Filter points within 1 std deviation
    mean_x = dfop.endX.mean()
    mean_y = dfop.endY.mean()
    std_x = dfop.endX.std()
    std_y = dfop.endY.std()
    
    # Create mask for points within 1 std dev
    mask = ((dfop.endX - mean_x)**2 / std_x**2 + (dfop.endY - mean_y)**2 / std_y**2) <= 1.5
    
    # Get points within 1 std dev
    points_within = np.column_stack([dfop.endY[mask], dfop.endX[mask]])
    
    # Create convex hull polygon
    if len(points_within) > 2:
        hull = ConvexHull(points_within)
        polygon = Polygon(points_within[hull.vertices], alpha=0.3, 
                         facecolor='None', edgecolor='blue', hatch='/////')
        ax.add_patch(polygon)
    
    ax.hlines(medX, xmin=-2, xmax=70, color='gray', linewidth=2, linestyle='--')
    ax.vlines(medY, ymin=-2, ymax=107, color='gray', linewidth=2, linestyle='--')

    arrow = FancyArrowPatch((70, medX), (70, 105),
                           arrowstyle='<->', 
                           linestyle = 'dashed',
                           mutation_scale=20,
                           linewidth=2,
                           color='gray')
    
    ax.add_patch(arrow)
    ax.text(72, (medX+105)/2, f'{(105-medX):.2f}m', color='gray', ha='center', va='center', rotation=90)

    ax.text(34, 2.75, '@adnaaan433', ha='center', va='center', alpha=0.75, fontfamily='serif', 
            bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=1))
    ax_text(34, 108, '<Passes Received>  |  <Key Pass Received>  |  <Assist Received>', ha='center', va='center', 
            highlight_textprops=[{'color':'#2680ff'},{'color':'#e322cc'},{'color':'green'}], ax=ax)
    ax.set_title('Pass Receiving Heatmap', fontsize=15, fontweight='bold', y=1.025)
    
    return
```

```python
def stats_counter(pname, team_name):
    dfp = df[(df['type'] == 'Pass') & (df['outcomeType'] == 'Successful') & (df['name'].shift(-1) == pname) & (df['teamName']==team_name)]
    dfkp = dfp[dfp['qualifiers'].str.contains('KeyPass')]
    dfas = dfp[dfp['qualifiers'].str.contains('IntentionalGoalAssist')]
    dfnt = dfp[dfp['endX'] >= 70]
    dfpen = dfp[(dfp['endX'] >= 87.5) & (dfp['endY'] >= 13.6) & (dfp['endY'] <= 54.6)]
    dfpro = dfp[(dfp['x'] >= 35) & (dfp['prog_pass'] >= 9.11)]
    dfxT = dfp[dfp['xT'] >= 0]
    dfcros = dfp[(dfp['qualifiers'].str.contains('Cross'))]
    dflb = dfp[(dfp['qualifiers'].str.contains('Longball')) & (dfp['x']<=52)]
    dfthr = dfp[dfp['qualifiers'].str.contains('Throughball')]
    
    pr_count = len(dfp)
    kp_count = len(dfkp)
    as_count = len(dfas)
    ft_count = len(dfnt)
    pen_count = len(dfpen)
    cros_count = len(dfcros)
    lb_count = len(dflb)
    thr_count = len(dfthr)
    
    return {
        'Name': pname,
        'Team Name': team_name,
        'Total Pass Received': pr_count,
        'KeyPass Received': kp_count,
        'Assist Received': as_count,
        'Final1/3 Pass Received': ft_count,
        'Pass Received in Pen. box': pen_count,
        'Cross Received': cros_count,
        'Longball Received': lb_count,
        'Pass Received In Behind': thr_count
    }

def player_dataset(df):
    # Get unique combinations of player and team
    dfpos = df[df['position'].isin(['FW'])]
    unique_pairs = dfpos[['name', 'teamName']].drop_duplicates()
    
    # Create a list of dictionaries to store the counts for each player-team combination
    data = []
    for _, row in unique_pairs.iterrows():
        pname = row['name']
        team_name = row['teamName']
        counts = stats_counter(pname, team_name)
        data.append(counts)
    
    # Convert the list of dictionaries to a DataFrame
    counts_df = pd.DataFrame(data)
    # Sort the DataFrame by 'xT Received' in descending order
    counts_df = counts_df.sort_values(by='Pass Received In Behind', ascending=False).reset_index(drop=True)

    return counts_df

counts_df = player_dataset(df)
```

```python
def create_player_ranking_viz(ax, df_stats, pname, team_name):
    # Define the stats columns (exclude Player column)
    stat_columns = [col for col in df_stats.columns if (col != 'Name' and col != 'Team Name')]
    
    # Calculate rankings for each stat (1 = best rank)
    rankings = {}
    player_values = {}
    stat_max_ranks = {}
    
    for stat in stat_columns:
        # For most stats, higher is better, but you might need to adjust this
        # based on your specific metrics
        df_sorted = df_stats.sort_values(stat, ascending=False)
        # df_sorted['rank'] = range(1, len(df_sorted) + 1)
        df_sorted['rank'] = df_sorted[stat].rank(method='dense', ascending=False).astype(int)
        
        # Get player's rank and actual value
        player_rank = df_sorted[(df_sorted['Name'] == pname) & (df_sorted['Team Name']==team_name)]['rank'].iloc[0]
        player_value = df_stats[(df_stats['Name'] == pname) & (df_stats['Team Name']==team_name)][stat].iloc[0]
        
        rankings[stat] = player_rank
        player_values[stat] = player_value

        # Track maximum rank for this specific stat
        stat_max_ranks[stat] = df_sorted['rank'].max()
    
    # Set up the plot
    y_positions = range(len(stat_columns))
    stat_labels = stat_columns
    
    # Create horizontal lines (bars)
    for i, stat in enumerate(stat_columns):
        ax.plot([0, 1], [i, i], color='#bfbfbf', linewidth=8, alpha=0.7)
    
    # Add scatter points for player rankings
    for i, stat in enumerate(stat_columns):
        rank = rankings[stat]
        value = player_values[stat]
        stat_max_rank = stat_max_ranks[stat]

        # Normalize the position: rank 1 should be at position 1.0, worst rank at position 0.0
        # This gives each bar its own independent scale
        normalized_pos = 1.0 - (rank - 1) / (stat_max_rank - 1) if stat_max_rank > 1 else 1.0
        
        # Plot the rank position (inverted so rank 1 is at the right)
        ax.scatter(normalized_pos, i, color='#66a6ff', s=250, zorder=5, edgecolor='k', linewidth=2)
        
        # Add the actual stat value as text
        ax.text(normalized_pos, i-0.25, f'{value}', fontsize=13, fontweight='bold', 
                bbox=dict(boxstyle='round,pad=0.3', facecolor='black', alpha=0.7),
                color='white', ha='center', va='center')
    
    # Customize the plot
    ax.set_yticks(y_positions)
    ax.set_yticklabels(stat_labels, fontsize=15, color='k')
    ax.yaxis.tick_right()
    ax.set_xlim(-0.025, 1.025)
    ax.set_ylim(-0.5, len(stat_columns) - 0.5)
    ax.invert_yaxis()
    
    # Remove x-axis as it's not needed for ranking visualization
    ax.set_xticks([])
    
    # Style the plot
    ax.set_facecolor('#f5f5f5')
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    
    # Add title
    plt.title(f'Pass Receiving Stats', fontsize=15, fontweight='bold', color='k', pad=20)
    
    plt.tight_layout()

    return 
```
