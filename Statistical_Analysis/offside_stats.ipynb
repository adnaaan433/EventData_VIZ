```python
import pandas as pd

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
```

```python
df = pd.read_csv(r"D:\FData\All_Top_5_League\2024_25\event_data\Top5_League_2024_25_event_data.csv")
```

```python
df = df[[
        'eventId', 'minute', 'second', 'type', 'outcomeType', 'x', 'y',
        'playerName', 'shirtNo', 'position', 'age', 'teamName',
        'oppositionTeamName', 'h_a', 'matchId', 'venueName',
]]
```

```python
df = df[df['type'].isin(['OffsideGiven', 'OffsidePass', 'OffsideProvoked'])]
```

```python
offside_provoked = df[df['type']=='OffsideProvoked']
offside_caught = df[df['type']=='OffsideGiven']
```

```python
df = df[[
        'eventId', 'minute', 'second', 'type', 'outcomeType', 'x', 'y',
        'playerName', 'shirtNo', 'position', 'age', 'teamName',
        'oppositionTeamName', 'h_a', 'matchId', 'venueName',
]]

df = df[df['type'].isin(['OffsideGiven', 'OffsidePass', 'OffsideProvoked'])]

offside_provoked = df[df['type']=='OffsideProvoked']
offside_caught = df[df['type']=='OffsideGiven']


```

```python
offside_provoked_summary = (
    offside_provoked
    .groupby(['matchId', 'teamName', 'oppositionTeamName'])
    .size()
    .reset_index(name='offsides_provoked')
    .sort_values(by='offsides_provoked', ascending=False)
    .reset_index(drop=True)
)
offside_provoked_summary = offside_provoked_summary[['teamName', 'offsides_provoked', 'oppositionTeamName']]
offside_provoked_summary = offside_provoked_summary.rename(columns={'oppositionTeamName': 'against'})

offside_provoked_summary.head(20)
```

```python
offside_caught_summary = (
    offside_caught
    .groupby(['matchId', 'playerName', 'oppositionTeamName'])
    .size()
    .reset_index(name='offsides_provoked')
    .sort_values(by='offsides_provoked', ascending=False)
    .reset_index(drop=True)
)
offside_caught_summary = offside_caught_summary[['playerName', 'offsides_provoked', 'oppositionTeamName']]
offside_caught_summary = offside_caught_summary.rename(columns={'oppositionTeamName': 'against'})

offside_caught_summary.head(20)
```
